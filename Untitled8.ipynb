```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("NYPD_Arrests_Data_(Historic)_20260509.csv")
```

```python
df=df.dropna(subset=["ARREST_BORO"])
```

```python
boroughs = df[df["ARREST_BORO"].isin(["B","M"])]
```

```python
counts = boroughs["ARREST_BORO"].value_counts()
```

```python
print("Number of arrests:")
print(counts)
```

```python
counts.plot(kind="bar")
plt.title("B VS M ARREST")
plt.xlabel("BOROUGH")
plt.ylabel("NUMBER OF ARRESTS")
plt.show()
```

```python
df['ARREST_DATE'] = pd.to_datetime(df['ARREST_DATE'], format='%m/%d/%Y', errors='coerce')

arrests_by_date = df.groupby('ARREST_DATE').size().reset_index(name='Arrest_Count')

plt.figure(figsize=(15, 7))
plt.plot(arrests_by_date['ARREST_DATE'], arrests_by_date['Arrest_Count'])
plt.title('Total Arrests Over Time')
plt.xlabel('Date')
plt.ylabel('Number of Arrests')
plt.grid(True)
plt.tight_layout()
plt.show()
```
